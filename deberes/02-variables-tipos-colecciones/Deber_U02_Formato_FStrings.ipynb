## 1. ¿Qué es un f-string?

Un **f-string** es un texto con una `f` antes de las comillas. Todo lo que pongas entre **llaves `{}`** se evalúa y su resultado se inserta en el texto:

```python
nombre = "Ana"
edad = 19
print(f"Me llamo {nombre} y tengo {edad} años")   # Me llamo Ana y tengo 19 años
```

Ventajas frente a unir con `+`:
- no hace falta convertir con `str()`;
- se leen igual que el texto final;
- dentro de las llaves puede ir **cualquier expresión**: `{edad + 1}`, `{precio * 2}`, `{len(nombre)}`.

> ⚠️ Si olvidas la `f`, Python muestra las llaves tal cual: `"Hola {nombre}"` → `Hola {nombre}`.

## 2. Formato de números

Después de la expresión se pone **`:`** y una **especificación de formato**:

| Formato | Significado | Ejemplo | Resultado |
|---|---|---|---|
| `{x:.2f}` | 2 decimales (redondea) | `f"{3.14159:.2f}"` | `3.14` |
| `{x:.0f}` | sin decimales | `f"{7.8:.0f}"` | `8` |
| `{x:,}` | separador de miles | `f"{1500000:,}"` | `1,500,000` |
| `{x:,.2f}` | miles y 2 decimales | `f"{1234.5:,.2f}"` | `1,234.50` |
| `{x:.1%}` | porcentaje (multiplica por 100) | `f"{0.256:.1%}"` | `25.6%` |
| `{x:+}` | muestra siempre el signo | `f"{5:+}"` | `+5` |

> El formato solo cambia **cómo se muestra** el número; la variable conserva su valor completo.

### 2.1 — Predice

## 3. Ancho y alineación

Puedes reservar un **ancho** fijo y decidir cómo se alinea el valor dentro de él:

| Formato | Significado | Ejemplo | Resultado (los `[ ]` muestran el ancho) |
|---|---|---|---|
| `{x:<10}` | a la izquierda, ancho 10 | `f"[{'ok':<6}]"` | `[ok    ]` |
| `{x:>10}` | a la derecha | `f"[{'ok':>6}]"` | `[    ok]` |
| `{x:^10}` | centrado | `f"[{'ok':^6}]"` | `[  ok  ]` |
| `{x:*^10}` | centrado, rellenando con `*` | `f"[{'ok':*^6}]"` | `[**ok**]` |
| `{n:05}` | entero con ceros a la izquierda | `f"{42:05}"` | `00042` |
| `{x:>8.2f}` | derecha, ancho 8, 2 decimales | `f"[{3.5:>8.2f}]"` | `[    3.50]` |

Por defecto, los **textos** se alinean a la izquierda y los **números** a la derecha.

### 3.1 — Predice
Los corchetes `[ ]` son parte del texto: sirven para **ver** los espacios.

````python
palabra = "hola"
valor = 7.5
print(f"[{palabra:<8}]")    # Mi predicción:
print(f"[{palabra:>8}]")    # Mi predicción:
print(f"[{palabra:^8}]")    # Mi predicción:
print(f"[{palabra:-^10}]")  # Mi predicción:
print(f"[{valor:>6}]")      # Mi predicción:
print(f"[{valor:<6.2f}]")   # Mi predicción:
print(f"[{palabra:2}]")     # Mi predicción:
print(f"{8:03}")            # Mi predicción:
````

## 4. La alternativa: `str.format()`

Antes de los f-strings se usaba `str.format()`. Las llaves quedan **vacías** (o con un formato) y los valores se pasan **en orden** dentro de `.format(...)`:

```python
nombre = "Ana"
promedio = 8.456
print("Estudiante: {} | Promedio: {:.2f}".format(nombre, promedio))
print(f"Estudiante: {nombre} | Promedio: {promedio:.2f}")   # mismo resultado
```

Los formatos (`.2f`, `>10`, `05`, `,` …) se escriben igual después de `:`.

### 4.1 — Traduce a `str.format()`
Escribe `version_format_1` y `version_format_2` con `str.format()` para que den **el mismo texto** que sus f-strings.

````python
ciudad = "Cuenca"
temperatura = 17.456
humedad = 0.82

version_f_1 = f"En {ciudad} hay {temperatura:.1f} °C"
version_f_2 = f"Humedad: {humedad:.0%} | Ciudad: {ciudad:>10}"

version_format_1 = None
version_format_2 = None

print(version_f_1)
print(version_format_1)
print(version_f_2)
print(version_format_2)
````

````python
# 🔎 Comprobación — no modifiques esta celda
assert version_format_1 == version_f_1 == "En Cuenca hay 17.5 °C", f"Obtuviste {version_format_1!r}"
assert version_format_2 == version_f_2 == "Humedad: 82% | Ciudad:     Cuenca", f"Obtuviste {version_format_2!r}"
print("✅ Ejercicio 4.1 correcto")
````
