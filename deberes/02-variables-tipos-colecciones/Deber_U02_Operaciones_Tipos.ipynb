## 3. Booleanos

Una **comparación** produce un valor `bool` (`True` o `False`):

| Operador | Significado | Ejemplo | Resultado |
|---|---|---|---|
| `==` | igual a | `5 == 5` | `True` |
| `!=` | distinto de | `5 != 5` | `False` |
| `<` `>` | menor / mayor que | `3 < 8` | `True` |
| `<=` `>=` | menor o igual / mayor o igual | `7 >= 7` | `True` |

> ⚠️ `=` **asigna** un valor; `==` **compara**.

Los booleanos se combinan con operadores lógicos:

| `a` | `b` | `a and b` | `a or b` | `not a` |
|---|---|---|---|---|
| `True` | `True` | `True` | `True` | `False` |
| `True` | `False` | `False` | `True` | `False` |
| `False` | `True` | `False` | `True` | `True` |
| `False` | `False` | `False` | `False` | `True` |

- `and`: es `True` solo si **ambos** son `True`.
- `or`: es `True` si **al menos uno** es `True`.
- `not`: invierte el valor.

### 3.1 — Predice

````python
print(5 > 3)            # Mi predicción:
print(5 == 5.0)         # Mi predicción:
print("5" == 5)         # Mi predicción:
print("abc" == "ABC")   # Mi predicción:
print("a" < "b")        # Mi predicción:
print("10" < "9")       # Mi predicción:
print(3 < 5 < 8)        # Mi predicción:
print(True and False)   # Mi predicción:
print(True or False)    # Mi predicción:
print(not (4 > 2))      # Mi predicción:
````

## 4. Operaciones entre mismos y distintos tipos

| Operación | Resultado | Ejemplo |
|---|---|---|
| `int` con `int` (`+ - * // % **`) | `int` | `7 + 2` → `9` |
| `int / int` | **siempre** `float` | `8 / 2` → `4.0` |
| `int` con `float` | `float` | `3 + 2.0` → `5.0` |
| `str + str` | `str` (concatena) | `"a" + "b"` → `"ab"` |
| `str * int` | `str` (repite) | `"a" * 3` → `"aaa"` |
| `str + int` | ❌ `TypeError` | `"a" + 1` |
| `bool` con `int` | `int` (`True` vale 1, `False` vale 0) | `True + 1` → `2` |

Si los tipos no son compatibles, **convierte** primero con `int()`, `float()` o `str()`.
