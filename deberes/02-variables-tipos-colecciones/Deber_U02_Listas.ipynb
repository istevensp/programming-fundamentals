## 1. ¿Qué es una lista?

Imagina que necesitas guardar las notas de 5 estudiantes. Con lo que sabes hasta ahora, harías esto:

```python

nota_2 = 7.0
nota_3 = 9.25
nota_4 = 6.5
nota_5 = 10.0
```

¿Y si fueran 40 estudiantes? Una **lista** guarda **varios valores en una sola variable**, en orden:

```python
notas = [8.5, 7.0, 9.25, 6.5, 10.0]
```

- Se escribe entre **corchetes `[ ]`** y los elementos se separan con **comas**.
- `len(lista)` dice **cuántos elementos** tiene.
- Una lista vacía se escribe `[]`.
- Una lista puede guardar cualquier tipo, pero lo normal es que todos sus elementos sean del **mismo tipo** (todas notas, todos nombres…).

```python
ciudades = ["Guayaquil", "Quito", "Cuenca"]
print(ciudades)          # ['Guayaquil', 'Quito', 'Cuenca']
print(len(ciudades))     # 3
print(type(ciudades))    # <class 'list'>
```

## 2. Acceder a los elementos: índices

Cada elemento tiene una **posición** llamada **índice**. Se cuenta **desde 0**. También se puede contar desde el final con índices **negativos**:

```python
ciudades = ["Guayaquil", "Quito", "Cuenca", "Manta", "Loja"]
```

| Elemento | `"Guayaquil"` | `"Quito"` | `"Cuenca"` | `"Manta"` | `"Loja"` |
|---|---|---|---|---|---|
| **Índice positivo** | `0` | `1` | `2` | `3` | `4` |

## 3. Modificar una lista

A diferencia de los textos, las listas **se pueden cambiar** después de crearlas.

| Operación | Qué hace | Ejemplo con `x = ["a", "b", "c"]` | `x` queda |
|---|---|---|---|
| `x[i] = valor` | **reemplaza** el elemento de la posición `i` | `x[1] = "z"` | `["a", "z", "c"]` |
| `x.append(valor)` | **agrega al final** | `x.append("d")` | `["a", "b", "c", "d"]` |
| `x.insert(i, valor)` | **inserta** en la posición `i` (los demás se corren) | `x.insert(0, "w")` | `["w", "a", "b", "c"]` |
| `x.remove(valor)` | **elimina la primera aparición** de ese valor | `x.remove("b")` | `["a", "c"]` |
| `x.pop(i)` | **saca** el elemento de la posición `i` y lo **devuelve** | `y = x.pop(0)` | `["b", "c"]` y `y` vale `"a"` |
| `x.pop()` | sin índice, saca el **último** | `y = x.pop()` | `["a", "b"]` y `y` vale `"c"` |

> `append`, `insert` y `remove` **modifican la lista** y no devuelven nada. Por eso **no** se escribe `x = x.append("d")`: eso guardaría `None` en `x`.

## 4. Operaciones con listas

| Operación | Qué hace | Ejemplo | Resultado |
|---|---|---|---|
| `lista + lista` | une dos listas en una **nueva** | `[1, 2] + [3]` | `[1, 2, 3]` |

## 5. Buscar elementos y valores repetidos

| Operación | Qué responde | Ejemplo con `r = ["A", "C", "A"]` | Resultado |
|---|---|---|---|
| `valor in lista` | ¿aparece? | `"A" in r` | `True` |
| `lista.count(valor)` | ¿cuántas veces aparece? | `r.count("A")` | `2` |
| `lista.index(valor)` | ¿en qué posición aparece **por primera vez**? | `r.index("A")` | `0` |
| `lista.index(valor, desde)` | primera aparición **a partir** de la posición `desde` | `r.index("A", 1)` | `2` |

> ⚠️ `index` solo encuentra la **primera** aparición. Si el valor está repetido, para encontrar la siguiente hay que buscar **desde** la posición que sigue.
>
> ⚠️ `index` produce `ValueError` si el valor **no está**. `in` y `count` nunca dan error: si no está, responden `False` y `0`.

## 7. Copia o alias: ¿la misma lista o otra?

Esto es **muy importante** y causa muchos errores:

```python
original = [1, 2, 3]
alias = original        # NO copia: es OTRO NOMBRE para la MISMA lista
copia = original[:]     # SÍ copia: la rebanada completa crea una lista NUEVA
```

Si modificas `original`, el cambio **se ve también desde `alias`**, porque son la misma lista. La `copia` no cambia.

> Con números y textos no pasa esto, porque no se pueden modificar: solo se reemplazan. Con listas sí, porque `append`, `remove`, `x[i] = …` **modifican la lista misma**.

### 7.1 — Predice

````python
original = [1, 2, 3]
alias = original
copia = original[:]

original.append(4)
alias[0] = 100

print(original)    # Mi predicción:
print(alias)       # Mi predicción:
print(copia)       # Mi predicción:
````

## 8. Listas y azar: librería `random`

| Función | Qué hace | ¿Puede repetir? | Devuelve |
|---|---|---|---|
| `random.choice(lista)` | elige **un** elemento | — | un elemento |
| `random.choices(lista, k=3)` | elige **3**, cada vez de entre **todos** | **sí** (con reemplazo) | una lista nueva |
| `random.sample(lista, k=3)` | elige **3 distintos** | **no** (sin reemplazo) | una lista nueva |
| `random.shuffle(lista)` | **desordena** la lista misma | — | `None` (modifica la original) |

> - `sample` no repite **posiciones**: si la lista ya tiene valores repetidos, el resultado puede tenerlos.
> - `shuffle` modifica la lista que recibe (como en la sección 7). Si necesitas conservar el orden original, **copia primero**. Por azar, el orden mezclado puede coincidir con el original.
> - Pedir con `sample` más elementos de los que hay produce `ValueError`.
