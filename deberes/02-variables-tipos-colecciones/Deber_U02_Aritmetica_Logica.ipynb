## 2. Operadores lógicos

Los operadores lógicos trabajan con valores `bool` y producen un `bool`.

| Nombre | Python | Símbolo | Es `True` cuando… |
|---|---|---|---|
| Conjunción | `p and q` | $p \land q$ | **ambos** son `True` |
| Disyunción | `p or q` | $p \lor q$ | **al menos uno** es `True` |
| Negación | `not p` | $\lnot p$ | `p` es `False` |

**Tabla de verdad**

| `p` | `q` | `p and q` | `p or q` | `not p` |
|---|---|---|---|---|
| `True` | `True` | `True` | `True` | `False` |
| `True` | `False` | `False` | `True` | `False` |
| `False` | `True` | `False` | `True` | `True` |
| `False` | `False` | `False` | `False` | `True` |

**Orden de evaluación completo** (de primero a último):

1. `**`
2. `* / // %`
3. `+ -`
4. comparaciones `== != < > <= >=`
5. `not`
6. `and`
7. `or`

Por eso `True or False and False` se evalúa como `True or (False and False)`. Ante la duda, **usa paréntesis**.
