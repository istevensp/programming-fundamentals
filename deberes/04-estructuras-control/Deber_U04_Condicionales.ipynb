## 1. Repaso: ¿qué es una condición?

Hasta ahora, tus programas ejecutaban **todas** las líneas, una tras otra. Con los condicionales, un programa puede **decidir** qué líneas ejecutar.

Para decidir, Python necesita una **condición**: una expresión que da `True` (verdadero) o `False` (falso). Ya las conoces de la Unidad 02:

| Operador | Significa | Ejemplo | Resultado |
|---|---|---|---|
| `==` | igual a | `5 == 5` | `True` |
| `!=` | distinto de | `5 != 3` | `True` |
| `>` / `<` | mayor / menor que | `7 < 3` | `False` |
| `>=` / `<=` | mayor o igual / menor o igual | `7 >= 7` | `True` |
| `in` | está contenido en | `"a" in "casa"` | `True` |
| `and` | las dos condiciones se cumplen | `5 > 3 and 2 > 8` | `False` |
| `or` | al menos una se cumple | `5 > 3 or 2 > 8` | `True` |
| `not` | invierte el resultado | `not (5 > 3)` | `False` |

⚠️ No confundas `=` (**guarda** un valor en una variable) con `==` (**pregunta** si dos valores son iguales).

## 2. `if`: hacer algo **solo si** se cumple una condición

```python
temperatura = 33

if temperatura > 30:                 # 1. if + condición + dos puntos
    print("Hace calor")              # 2. el BLOQUE va con sangría (4 espacios)
    print("Toma agua")

print("Fin del reporte")             # 3. sin sangría: ya está FUERA del if
```

| Parte | Qué es |
|---|---|
| `if` | palabra clave: "si…" |
| `temperatura > 30` | la condición (da `True` o `False`) |
| `:` | los dos puntos son **obligatorios** |
| bloque con sangría | las líneas que se ejecutan **solo si** la condición es `True` |

Cómo lo ejecuta Python:

```text
            ┌──── True ───► ejecuta el bloque del if ──┐
condición ──┤                                          ├──► sigue con el programa
            └──── False ───────────────────────────────┘
```

**La sangría decide qué está dentro del `if`.** Todas las líneas con la misma sangría debajo del `if` forman su bloque. La primera línea que vuelve al margen izquierdo ya está fuera y se ejecuta **siempre**.

## 4. `if-elif-else`: varios caminos

Cuando hay **más de dos** posibilidades, se agregan condiciones con `elif` ("si no, si…"):

```python
nota = 8.7

if nota >= 9:
    categoria = "Excelente"
elif nota >= 8:
    categoria = "Muy bueno"
elif nota >= 6:
    categoria = "Aprobado"
else:
    categoria = "Reprobado"
```

Reglas:
1. Python revisa las condiciones **de arriba hacia abajo**.
2. Ejecuta **solo el primer** bloque cuya condición sea `True` y se **salta todos los demás**, aunque también se cumplan.
3. `elif` sí lleva condición; `else` no, y es **opcional**: si no está y nada se cumple, no se ejecuta ningún bloque.

Con `nota = 8.7`: `8.7 >= 9` es `False` → `8.7 >= 8` es `True` → `categoria = "Muy bueno"` y ya no revisa nada más.

## 5. Condiciones compuestas: `and`, `or`, `not`

Una condición puede combinar varias preguntas:

| Operador | Es `True` cuando… | Ejemplo |
|---|---|---|
| `a and b` | se cumplen **las dos** | `edad >= 18 and tiene_entrada` |
| `a or b` | se cumple **al menos una** | `dia == "sábado" or dia == "domingo"` |
| `not a` | `a` **no** se cumple | `not llueve` |

- Para un rango, Python permite encadenar: `0 <= nota <= 10` es lo mismo que `nota >= 0 and nota <= 10`.
- Python evalúa `not` primero, luego `and` y al final `or`. Cuando mezcles `and` y `or`, **usa paréntesis** para que se lea lo que quieres decir.
- Cada lado de `and` / `or` debe ser una condición completa: `dia == "sábado" or "domingo"` **no** pregunta si el día es domingo (ver 5.3).

### 5.1 — Predice

````python
def puede_ingresar(edad, tiene_entrada):
    if edad >= 18 and tiene_entrada:
        return "Pasa"
    else:
        return "No pasa"

def precio_bus(es_estudiante, edad):
    if es_estudiante or edad >= 65:
        return "Media tarifa"
    else:
        return "Tarifa completa"

def consejo(llueve):
    if not llueve:
        return "Sal sin paraguas"
    else:
        return "Lleva paraguas"

print(puede_ingresar(20, True))     # Mi predicción:
print(puede_ingresar(20, False))    # Mi predicción:
print(puede_ingresar(15, True))     # Mi predicción:
print(precio_bus(True, 20))         # Mi predicción:
print(precio_bus(False, 70))        # Mi predicción:
print(precio_bus(False, 30))        # Mi predicción:
print(consejo(False))               # Mi predicción:
````
