````python
# 🔧 Celda de preparación — ejecútala primero y NO la modifiques.
# Define tres ayudas que usan las celdas de comprobación:
#   usa_variables_externas(f): lista las variables de afuera que usa la función f (debe quedar vacía)
#   salida_de(f, ...): devuelve lo que la función f imprime en pantalla
#   usa(f, nombre): indica si la función f usa una función de Python (sum…) o un método (.replace…)
import inspect, io, contextlib

def usa_variables_externas(funcion):
    externas = inspect.getclosurevars(funcion).globals
    return [nombre for nombre in externas
            if not callable(externas[nombre]) and not inspect.ismodule(externas[nombre])]

def salida_de(funcion, *argumentos):
    pantalla = io.StringIO()
    with contextlib.redirect_stdout(pantalla):
        funcion(*argumentos)
    return pantalla.getvalue()

def usa(funcion, nombre):
    variables = inspect.getclosurevars(funcion)
    return nombre in variables.builtins or nombre in variables.unbound

print("✅ Preparación lista")
````

## 1. Calentamiento

Casi todos los ejercicios de este deber siguen uno de estos moldes:

| Molde | Antes del `for` | Dentro del `for` | Después |
|---|---|---|---|
| **Contar** los que cumplen | `cantidad = 0` | `if condición: cantidad = cantidad + 1` | usar `cantidad` |
| **Acumular** los que cumplen | `total = 0` | `if condición: total = total + valor` | usar `total` |
| **Filtrar** | `resultado = []` | `if condición: resultado.append(valor)` | devolver `resultado` |
| **Transformar** según el caso | `resultado = []` | `if …: append(a)` / `else: append(b)` | devolver `resultado` |
| **Clasificar** en varios grupos | varios contadores en 0 | `if` / `elif` / `else` | devolver los contadores |
| **¿Hay alguno? / ¿todos?** | `cantidad = 0` | contar los que cumplen | `cantidad > 0` / `cantidad == len(...)` |

Las decisiones van **dentro** del `for` (una por vuelta). El `return` del resultado va **después** del `for`, con la misma sangría que el `for`.

texto = input("Edad: ").strip()                       # primer intento
for intento in range(2):                      # hasta 2 reintentos
    if not es_entero(texto):                  # solo si sigue siendo inválido...

        texto = input("Edad: ").strip()               # ...se vuelve a pedir
if es_entero(texto):                          # después del ciclo se decide

else:
