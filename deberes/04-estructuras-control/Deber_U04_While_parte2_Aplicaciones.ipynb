````python
# 🔧 Celda de preparación — ejecútala primero y NO la modifiques.
# Define las ayudas que usan las celdas de comprobación:
#   usa_variables_externas(f): lista las variables de afuera que usa la función f (debe quedar vacía)
#   salida_de(f, ...): devuelve lo que la función f imprime en pantalla
#   usa(f, nombre): indica si la función f usa una función de Python (sum, max…) o un método (.count…)
#   usa_for(f): indica si la función f tiene un ciclo for (en este deber los ciclos van con while)
import inspect, io, contextlib, dis

def usa_variables_externas(funcion):
    externas = inspect.getclosurevars(funcion).globals
    return [nombre for nombre in externas
            if not callable(externas[nombre]) and not inspect.ismodule(externas[nombre])]

def salida_de(funcion, *argumentos):
    pantalla = io.StringIO()
    with contextlib.redirect_stdout(pantalla):
        funcion(*argumentos)
    return pantalla.getvalue()

def usa(funcion, nombre):
    variables = inspect.getclosurevars(funcion)
    return nombre in variables.builtins or nombre in variables.unbound

def usa_for(funcion):
    def tiene_for(codigo):
        return (any(instruccion.opname == "FOR_ITER" for instruccion in dis.get_instructions(codigo))
                or any(tiene_for(c) for c in codigo.co_consts if inspect.iscode(c)))
    return tiene_for(funcion.__code__)

print("✅ Preparación lista")
````

## 1. ¿`for` o `while`?

Los dos repiten, pero sirven para cosas distintas. La pregunta clave es: **¿sé cuántas vueltas habrá antes de empezar?**

| Usa `for` cuando… | Usa `while` cuando… |
|---|---|
| recorres **todos** los elementos de una lista o texto | quieres **detenerte antes** del final (al encontrar algo) |
| la cantidad de vueltas es fija: `range(n)` | la cantidad depende de **lo que pase** en el ciclo |
| | repites **hasta** que el usuario escriba algo válido o un valor de salida |
| | repites **hasta** alcanzar una meta (saldo, puntaje, precisión) |

Todo `for` puede escribirse con `while` (lo hiciste en la Parte 1), pero no al revés: un `for` no puede "esperar" a que el usuario se equivoque una cantidad desconocida de veces.

### Repaso: `and` y `or` en la condición
Python evalúa `and` y `or` **de izquierda a derecha** y se detiene en cuanto conoce el resultado:

| Expresión | Si la parte izquierda es… | ¿Evalúa la derecha? |
|---|---|---|
| `A and B` | falsa | **no**: el resultado ya es falso |
| `A or B` | verdadera | **no**: el resultado ya es verdadero |

Por eso `posicion < len(lista) and lista[posicion] != x` nunca da `IndexError`: si la posición se salió, no se lee `lista[posicion]`.

### 1.1 — Predice

````python
datos = [5, 3, 8, 1]
i = 0
while i < len(datos) and datos[i] != 8:
    i = i + 1
print(i)                       # Mi predicción:

n = 100
pasos = 0
while n > 1:
    n = n // 3
    pasos = pasos + 1
print(n, pasos)                # Mi predicción:

a = 0
b = 10
while a < 5 or b > 8:
    a = a + 2
    b = b - 1
print(a, b)                    # Mi predicción:
````

### Máximo común divisor (algoritmo de Euclides)
Para hallar el MCD de `a` y `b`: **mientras `b` no sea 0**, reemplaza `(a, b)` por `(b, a % b)`. Cuando `b` llega a 0, el MCD es `a`.

| Vuelta | `a` | `b` | `a % b` |
|---|---|---|---|
| inicio | 48 | 18 | 12 |
| 1 | 18 | 12 | 6 |
| 2 | 12 | 6 | 0 |
| 3 | 6 | 0 | → termina: MCD = 6 |

> Para reemplazar los dos valores a la vez, guarda el residuo en una variable auxiliar: `residuo = a % b`, luego `a = b` y `b = residuo`.

### Números primos
Un número es **primo** si es mayor que 1 y solo es divisible entre 1 y sí mismo. Para saberlo, se prueban divisores desde 2 **mientras** no se haya encontrado uno. Basta probar hasta la raíz cuadrada: si `divisor * divisor > n` y no apareció ninguno, ya no aparecerá.

```python
divisor = 2
while divisor * divisor <= n and n % divisor != 0:
    divisor = divisor + 1
# si salió porque divisor * divisor > n, no encontró divisores: es primo
```

### Búsqueda binaria
Para buscar en una lista **ordenada** no hace falta revisar uno por uno: se mira el elemento **del medio** y se descarta la mitad donde no puede estar, como al buscar una palabra en el diccionario. Se usan dos posiciones, `inicio` y `fin`, que encierran la zona donde todavía puede estar:

```text
Buscar 23 en [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
vuelta 1: inicio=0 fin=9 medio=4 → 16 < 23: descarto la izquierda  → inicio = 5
vuelta 2: inicio=5 fin=9 medio=7 → 56 > 23: descarto la derecha    → fin = 6
vuelta 3: inicio=5 fin=6 medio=5 → 23 == 23: encontrado en la posición 5
```

El ciclo sigue **mientras** la zona no esté vacía (`inicio <= fin`) **y** el del medio no sea el buscado:

```python
inicio = 0
fin = len(lista) - 1
medio = (inicio + fin) // 2
while inicio <= fin and lista[medio] != buscado:
    if lista[medio] < buscado:
        inicio = medio + 1
    else:
        fin = medio - 1
    medio = (inicio + fin) // 2
```

En una lista de 1000 elementos, la búsqueda uno por uno puede necesitar 1000 vueltas; la binaria, **como máximo 10**, porque cada vuelta parte la zona por la mitad.
