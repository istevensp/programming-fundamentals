````python
# 🔧 Celda de preparación — ejecútala primero y NO la modifiques.
# Define las ayudas que usan las celdas de comprobación:
#   usa_variables_externas(f): lista las variables de afuera que usa la función f (debe quedar vacía)
#   salida_de(f, ...): devuelve lo que la función f imprime en pantalla
#   usa(f, nombre): indica si la función f usa una función de Python (sum, max…) o un método (.count…)
#   usa_for(f): indica si la función f tiene un ciclo for (en este deber los ciclos van con while)
import inspect, io, contextlib, dis

def usa_variables_externas(funcion):
    externas = inspect.getclosurevars(funcion).globals
    return [nombre for nombre in externas
            if not callable(externas[nombre]) and not inspect.ismodule(externas[nombre])]

def salida_de(funcion, *argumentos):
    pantalla = io.StringIO()
    with contextlib.redirect_stdout(pantalla):
        funcion(*argumentos)
    return pantalla.getvalue()

def usa(funcion, nombre):
    variables = inspect.getclosurevars(funcion)
    return nombre in variables.builtins or nombre in variables.unbound

def usa_for(funcion):
    def tiene_for(codigo):
        return (any(instruccion.opname == "FOR_ITER" for instruccion in dis.get_instructions(codigo))
                or any(tiene_for(c) for c in codigo.co_consts if inspect.iscode(c)))
    return tiene_for(funcion.__code__)

print("✅ Preparación lista")
````

## 1. ¿Para qué otro ciclo?

Con `for` repites algo **una vez por cada elemento**: sabes de antemano cuántas vueltas habrá (los elementos de la lista, los números del `range`). Pero muchas veces **no sabes cuántas vueltas harán falta**:

| Situación | ¿Cuántas vueltas? |
|---|---|
| Ahorrar $20 por mes hasta juntar $100 | depende de la meta y del aporte |
| Pedir la edad **hasta** que el usuario escriba una válida | depende de cuántas veces se equivoque |
| Sumar números **hasta** que escriba 0 | depende del usuario |
| Buscar un nombre en una lista y **parar** al encontrarlo | depende de dónde esté |

Para eso existe `while` ("mientras"): repite su bloque **mientras su condición sea verdadera**.

```python
saldo = 0                      # 1. estado inicial
meta = 100
aporte = 20

while saldo < meta:            # 2. condición: ¿todavía no llego?
    saldo = saldo + aporte     # 3. actualización: me acerco a la meta
    print(saldo)

print("Meta alcanzada")
```

```text
20
40
60
80
100
Meta alcanzada
```

Todo `while` tiene **tres partes**. Antes de escribir uno, identifícalas:

| Parte | Pregunta | En el ejemplo |
|---|---|---|
| **Estado inicial** | ¿con qué valores empiezo? | `saldo = 0` |
| **Condición** | ¿cuándo debo **seguir** repitiendo? | `saldo < meta` |
| **Actualización** | ¿qué cambia en cada vuelta para **acercarme al final**? | `saldo = saldo + aporte` |

Cómo lo ejecuta Python:

```text
        ┌──────────────────────────────────────┐
        ▼                                      │
¿la condición es verdadera? ── sí ──► ejecuta el cuerpo
        │
        no
        ▼
sigue con el programa
```

Fíjate: la condición se revisa **antes de cada vuelta**, también antes de la primera. Cuando `saldo` llega a 100, `100 < 100` es falso y el ciclo termina.

## 2. Contar con `while`

El uso más básico de `while` es **contar**: una variable empieza en un valor y se actualiza en cada vuelta hasta llegar al final. Es lo mismo que hace `for` con `range`, pero escribiendo **tú** las tres partes:

| Con `for` | Con `while` |
|---|---|
| `for numero in range(1, 6):`<br>`    print(numero)` | `numero = 1`<br>`while numero <= 5:`<br>`    print(numero)`<br>`    numero = numero + 1` |
| `range` crea y avanza la variable por ti | tú la **creas** antes, la **comparas** en la condición y la **avanzas** al final del cuerpo |

Para contar **de 2 en 2**, la actualización suma 2; para contar **hacia atrás**, resta:

```python
numero = 10
while numero > 0:
    print(numero)
    numero = numero - 3
```

```text
10
7
4
1
```

Y si la condición es **falsa desde el principio**, el cuerpo **no se ejecuta ni una vez**:

```python
numero = 10
while numero < 5:      # 10 < 5 es falso: no entra
    print(numero)
print("Nunca entré")
```

## 3. Trazas y errores típicos

### Tabla de traza
Para entender un `while` sin ejecutarlo, haz una **tabla de traza**: una fila por vuelta, con el valor de cada variable **al final** de la vuelta y el resultado de la condición.

```python
total = 0
numero = 1
while total < 10:
    total = total + numero
    numero = numero + 2
print(total, numero)
```

| Vuelta | `total` | `numero` | ¿`total < 10`? |
|---|---|---|---|
| inicio | 0 | 1 | sí → entra |
| 1 | 1 | 3 | sí → otra vuelta |
| 2 | 4 | 5 | sí → otra vuelta |
| 3 | 9 | 7 | sí → otra vuelta |
| 4 | 16 | 9 | **no** → termina |

Se muestra `16 9`.

### Los cuatro errores típicos

| Error | Ejemplo | Qué pasa |
|---|---|---|
| **Olvidar la actualización** | `i = 0`<br>`while i < 5:`<br>`    print(i)` | `i` siempre vale 0: **ciclo infinito** |
| **Actualizar en la dirección equivocada** | `i = 10`<br>`while i > 0:`<br>`    i = i + 1` | `i` se aleja del final: **ciclo infinito** |
| **Condición falsa desde el inicio** | `i = 10`<br>`while i < 5:` | el cuerpo **nunca** se ejecuta |
| **Pasarse o quedarse corto por uno** | `while i < 5` cuando debía ser `<=` | hace **una vuelta de menos** (o de más) |

> **Cómo prevenir un ciclo infinito:** antes de ejecutar, pregúntate *"¿qué línea del cuerpo acerca la condición a ser falsa?"*. Si no la encuentras, el ciclo no terminará. Si igual te pasa, detén la celda con **■**.

### Dígitos de un número
Con `//` y `%` entre 10 se separa un número entero en sus dígitos, **de derecha a izquierda**:

| Operación | Resultado con `n = 2026` | Qué hace |
|---|---|---|
| `n % 10` | `6` | da el **último** dígito |
| `n // 10` | `202` | **quita** el último dígito |

Repitiendo `n = n // 10` **mientras `n > 0`**, se recorren todos los dígitos sin saber cuántos son:

| Vuelta | `n % 10` | `n` después de `n = n // 10` |
|---|---|---|
| inicio | | 2026 |
| 1 | 6 | 202 |
| 2 | 2 | 20 |
| 3 | 0 | 2 |
| 4 | 2 | 0 → termina |

## 5. Recorrer listas con `while` y condición compuesta

Una lista también se recorre con `while`, usando una **posición** que empieza en 0 y avanza de 1 en 1 **mientras** sea válida:

```python
colores = ["rojo", "verde", "azul"]
posicion = 0
while posicion < len(colores):
    print(posicion, colores[posicion])
    posicion = posicion + 1
```

La ventaja aparece cuando quieres **detenerte antes del final** (por ejemplo, al encontrar algo). Como no se puede usar `break`, se agrega la razón para seguir **a la condición**, con `and`:

```python
posicion = 0
while posicion < len(numeros) and numeros[posicion] != buscado:
    posicion = posicion + 1
# al salir: o se acabó la lista (posicion == len(numeros)) o numeros[posicion] es el buscado
```

Se lee: *"avanza **mientras** queden elementos **y** el actual no sea el buscado"*.

> ⚠️ **El orden de la condición importa.** Python evalúa `and` de izquierda a derecha y, si la primera parte es falsa, **no evalúa la segunda**. Por eso primero va `posicion < len(numeros)`: cuando la posición se sale de la lista, la condición se detiene **antes** de intentar leer `numeros[posicion]`, que daría `IndexError`. Si las inviertes, el programa falla cuando el valor no está.

| Vuelta | `posicion` | `numeros[posicion]` | ¿sigue? (buscado = 7, `numeros = [4, 9, 7, 2]`) |
|---|---|---|---|
| inicio | 0 | 4 | `0 < 4` y `4 != 7` → sí |
| 1 | 1 | 9 | `1 < 4` y `9 != 7` → sí |
| 2 | 2 | 7 | `2 < 4` y `7 != 7` → **no**: encontrado en la posición 2 |

## 6. Condiciones con funciones booleanas (sin banderas)

Una **bandera** es una variable `True`/`False` que se usa solo para "encender" o "apagar" un ciclo:

```python
# ❌ Con bandera (desaconsejado)
encontrado = False
posicion = 0
while not encontrado and posicion < len(nombres):
    if nombres[posicion] == "Eva":
        encontrado = True
    else:
        posicion = posicion + 1
```

Funciona, pero para entender **por qué** se repite hay que seguir la bandera por todo el código. Es mejor que la condición diga directamente la razón:

```python
# ✅ La condición dice por qué se repite
posicion = 0
while posicion < len(nombres) and nombres[posicion] != "Eva":
    posicion = posicion + 1
```

Cuando la condición se vuelve larga, se le pone **nombre** con una **función booleana** (una función que devuelve `True` o `False`):

```python
def puede_seguir(acumulado, limite, incremento):
    return acumulado + incremento <= limite


total = 0
while puede_seguir(total, 25, 6):
    total = total + 6
print(total)   # 24: sumar otro 6 daría 30, que pasa de 25
```

| Con bandera | Con condición o función booleana |
|---|---|
| hay que leer todo el cuerpo para saber cuándo termina | la línea del `while` dice cuándo termina |
| una variable más que mantener | ninguna variable extra |
| fácil olvidar cambiarla (ciclo infinito) | la condición depende de los datos reales |

> La función booleana **recibe por parámetros** todo lo que necesita y devuelve directamente la comparación: `return a + b <= c`, no `if a + b <= c: return True else: return False`.
