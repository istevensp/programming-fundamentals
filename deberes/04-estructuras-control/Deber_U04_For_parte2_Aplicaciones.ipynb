````python
# 🔧 Celda de preparación — ejecútala primero y NO la modifiques.
# Define tres ayudas que usan las celdas de comprobación:
#   usa_variables_externas(f): lista las variables de afuera que usa la función f (debe quedar vacía)
#   salida_de(f, ...): devuelve lo que la función f imprime en pantalla
#   usa(f, nombre): indica si la función f usa una función de Python (max, min…) o un método (.index…)
import inspect, io, contextlib

def usa_variables_externas(funcion):
    externas = inspect.getclosurevars(funcion).globals
    return [nombre for nombre in externas
            if not callable(externas[nombre]) and not inspect.ismodule(externas[nombre])]

def salida_de(funcion, *argumentos):
    pantalla = io.StringIO()
    with contextlib.redirect_stdout(pantalla):
        funcion(*argumentos)
    return pantalla.getvalue()

def usa(funcion, nombre):
    variables = inspect.getclosurevars(funcion)
    return nombre in variables.builtins or nombre in variables.unbound

print("✅ Preparación lista")
````

## 3. Empates: cuando hay más de un ganador

`.index()` y el patrón del candidato devuelven **una sola** posición. Si el problema pide **todas**, se hace en dos pasos:

1. obtener el valor mayor (o menor);
2. recorrer con `enumerate` y guardar **cada** posición donde aparece ese valor.

```python
ventas = [120, 180, 95, 180]
mayor_venta = max(ventas)
posiciones = []
for posicion, venta in enumerate(ventas):
    if venta == mayor_venta:
        posiciones.append(posicion)
print(mayor_venta, posiciones)       # 180 [1, 3]
```

| El enunciado pide… | Usa |
|---|---|
| "el mayor" (solo el valor) | `max(lista)` |
| "la posición del mayor" (la primera) | `lista.index(max(lista))` |

## 4. Buscar sin banderas

Una **bandera** es una variable `True`/`False` que se "enciende" dentro de un ciclo para recordar que algo pasó:

```python
hay_negativo = False              # ❌ bandera
for numero in numeros:
    if numero < 0:
        hay_negativo = True
```

Funciona, pero Coordinación pide **evitarlas**: esconden la pregunta real y, con `while`, llevan a ciclos difíciles de seguir. Hay formas más claras de responder "¿hay alguno…?" o "¿todos…?":

| Pregunta | Sin bandera |
|---|---|
| ¿está este valor? | `valor in lista` |
| ¿hay **alguno** que cumpla? | contar los que cumplen y preguntar `cantidad > 0` |
| ¿**todos** cumplen? | contar los que cumplen y preguntar `cantidad == len(lista)` (o contar los que **no** cumplen y preguntar `== 0`) |
| ¿dónde está el **primero** que cumple? | guardar las posiciones que cumplen y tomar la primera |

Y todo esto dentro de una **función booleana** con un nombre que diga la pregunta: `hay_negativos(numeros)`.

## 7. Comprensiones de listas (solo si las puedes explicar)

Muchos filtros y transformaciones de la Parte 1 siguen el mismo molde: lista vacía, `for`, (a veces) `if`, `append`. Python permite escribir ese molde en **una línea**, llamada **comprensión de lista**:

```python
# Ciclo tradicional
cuadrados_pares = []
for numero in range(1, 7):
    if numero % 2 == 0:
        cuadrados_pares.append(numero ** 2)

# Comprensión equivalente
cuadrados_pares = [numero ** 2 for numero in range(1, 7) if numero % 2 == 0]
```

Se lee: "**para cada** número del 1 al 6, **si** es par, guardar su cuadrado".

| Parte de la comprensión | En el ciclo tradicional |
|---|---|
| `numero ** 2` (al inicio) | lo que va dentro del `append(...)` |
| `for numero in range(1, 7)` | la línea del `for` |
| `if numero % 2 == 0` (opcional, al final) | el `if` |

⚠️ Regla del curso: usa una comprensión **solo si puedes explicarla con palabras**. Si no, escribe el ciclo tradicional: es igual de correcto.
