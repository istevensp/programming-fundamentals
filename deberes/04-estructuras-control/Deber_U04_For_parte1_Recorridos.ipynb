````python
# 🔧 Celda de preparación — ejecútala primero y NO la modifiques.
# Define tres ayudas que usan las celdas de comprobación:
#   usa_variables_externas(f): lista las variables de afuera que usa la función f (debe quedar vacía)
#   salida_de(f, ...): devuelve lo que la función f imprime en pantalla
#   usa(f, nombre): indica si la función f usa una función de Python (sum, max…) o un método (.count…)
import inspect, io, contextlib

def usa_variables_externas(funcion):
    externas = inspect.getclosurevars(funcion).globals
    return [nombre for nombre in externas
            if not callable(externas[nombre]) and not inspect.ismodule(externas[nombre])]

def salida_de(funcion, *argumentos):
    pantalla = io.StringIO()
    with contextlib.redirect_stdout(pantalla):
        funcion(*argumentos)
    return pantalla.getvalue()

def usa(funcion, nombre):
    variables = inspect.getclosurevars(funcion)
    return nombre in variables.builtins or nombre in variables.unbound

print("✅ Preparación lista")
````

## 1. ¿Para qué sirve un ciclo?

Imagina que debes mostrar los números del 1 al 5. Sin ciclos:

```python
print(1)
print(2)
print(3)
print(4)
print(5)
```

¿Y si fueran 1000? Un **ciclo** repite un bloque de código **sin escribirlo muchas veces**. El ciclo `for` repite el bloque **una vez por cada elemento** de una colección:

```python
for numero in [1, 2, 3, 4, 5]:     # para cada número de la lista...
    print(numero)                  # ...ejecuta este bloque
```

| Parte | Qué es |
|---|---|
| `for` | palabra clave: "para cada…" |
| `numero` | la **variable del ciclo**: en cada vuelta toma el **siguiente** elemento |
| `in` | "en…" |
| `[1, 2, 3, 4, 5]` | lo que se recorre: una lista, un texto o un `range` |
| `:` | los dos puntos son **obligatorios** |
| bloque con sangría | el **cuerpo**: se ejecuta **una vez por elemento** |

Cada repetición se llama **iteración** (o vuelta). Cómo lo ejecuta Python:

```text
        ┌──────────────────────────────────────────────┐
        ▼                                              │
¿quedan elementos? ── sí ──► numero = el siguiente ──► ejecuta el cuerpo
        │
        no
        ▼
sigue con el programa
```

| Forma | Genera | Ejemplo | Resultado |

## 4. Acumuladores: sumar, multiplicar y unir dentro de un ciclo

Un **acumulador** es una variable que va **juntando** un resultado vuelta a vuelta. Siempre tiene tres pasos:

```python
def sumar_lista(numeros):
    total = 0                      # 1. INICIALIZAR antes del ciclo
    for numero in numeros:
        total = total + numero     # 2. ACTUALIZAR dentro del ciclo
    return total                   # 3. USAR después del ciclo
```

Una **traza** muestra cómo cambian las variables en cada vuelta. Para `sumar_lista([4, 7, 2])`:

| Vuelta | `numero` | `total` antes | `total` después |
|---|---|---|---|
| (antes del ciclo) | — | — | 0 |
| 1 | 4 | 0 | 4 |
| 2 | 7 | 4 | 11 |
| 3 | 2 | 11 | 13 |

Resultado: `13`.

| Para… | Valor inicial | Actualización |
|---|---|---|
| sumar | `0` | `total = total + valor` |
| multiplicar | `1` (con 0, todo daría 0) | `producto = producto * valor` |
| unir textos | `""` (texto vacío) | `texto = texto + parte` |
| construir una lista | `[]` (lista vacía) | `lista.append(valor)` (sección 7) |

## 5. Contadores y filtros: `if` dentro del `for`

Un **contador** es un acumulador que suma **1** cada vez que se cumple una condición:

```python
def contar_pares(numeros):
    cantidad = 0                       # inicializar
    for numero in numeros:
        if numero % 2 == 0:            # ¿este elemento cumple?
            cantidad = cantidad + 1    # contar
    return cantidad
```

Un **filtro** construye una lista nueva solo con los elementos que cumplen una condición, usando `append` (lo viste en la Unidad 02):

```python
def filtrar_pares(numeros):
    pares = []                         # lista vacía
    for numero in numeros:
        if numero % 2 == 0:
            pares.append(numero)       # se agrega solo si cumple
    return pares
```

## 6. `enumerate`: posición y valor a la vez

A veces necesitas **la posición** de cada elemento, no solo su valor: para numerar una lista, para buscar dónde está algo o para usar **listas paralelas**. `enumerate` te da las dos cosas en cada vuelta:

```python
notas = [8.5, 7.2, 9.1]

for posicion, nota in enumerate(notas):
    print(posicion, nota)
```

```text
0 8.5
1 7.2
2 9.1
```

| Forma | Cuándo usarla |
|---|---|
| `for nota in notas:` | solo necesitas el valor |
| `for posicion, nota in enumerate(notas):` | necesitas la posición **y** el valor |
| `for posicion in range(len(notas)):` | funciona, pero con `enumerate` se lee mejor |

Las posiciones empiezan en **0**, como los índices. Para numerar desde 1 al mostrar, usa `posicion + 1`.
