## 1. ¿Qué es una función?

Una **función** es un bloque de código con **nombre** que hace una tarea. Se escribe una vez y se usa **cuantas veces quieras**.

¡Ya has usado muchas! `print`, `len`, `input`, `int`, `type`… son funciones que vienen con Python:

```python
largo = len("Python")
#  ↑      ↑     ↑
#  |      |     └─ argumento: el dato que le das
#  |      └─ nombre de la función
#  └─ el resultado que la función DEVUELVE se guarda en una variable
```

Piensa en una función como una **máquina**: le das datos (**argumentos**), hace su trabajo y te **devuelve** un resultado.

Algunas funciones **devuelven** un valor (como `len`) y otras solo **hacen algo** (como `print`, que muestra en pantalla y no devuelve nada útil).

### ¿Por qué crear nuestras propias funciones?

Mira este programa: imprime el recibo de tres compras.

```python
print("--------------------")
print("Producto: Cuaderno")
print("Total: $5.00")
print("--------------------")
print("Producto: Esfero")
print("Total: $0.90")
print("--------------------")
print("Producto: Regla")
print("Total: $0.80")
```

Hay un **patrón que se repite**: separador, producto y total. Si mañana quieres cambiar el separador, tendrías que hacerlo en muchos lugares (y es fácil olvidar uno). Con una función, el patrón se escribe **una sola vez**:

```python
def mostrar_recibo(producto, total):
    print("--------------------")
    print(f"Producto: {producto}")
    print(f"Total: ${total:.2f}")

mostrar_recibo("Cuaderno", 5)
mostrar_recibo("Esfero", 0.9)
mostrar_recibo("Regla", 0.8)
```

Las funciones sirven para:
1. **No repetir código.**
2. **Darle nombre a una tarea**, para que el programa se lea como una lista de pasos (`calcular_iva`, `mostrar_recibo`…).
3. **Dividir un problema grande** en partes pequeñas que se pueden probar por separado.

### Vocabulario

## 2. Crear y llamar tus propias funciones

```python
def saludar():                       # 1. def + nombre + () + dos puntos
    print("¡Hola!")                  # 2. el CUERPO va con sangría (4 espacios)
    print("Bienvenido")

saludar()                            # 3. LLAMAR a la función: nombre + ()
```

- **Definir** la función (`def ...`) **no la ejecuta**: solo la guarda con ese nombre.

- Sin los paréntesis, `saludar` es solo el nombre: **no** la ejecuta.

## 4. `return`: devolver un resultado

Hasta ahora tus funciones **mostraban** cosas. Pero casi siempre queremos que una función **calcule y entregue** un resultado para **seguir usándolo**:

```python
def duplicar(numero):
    return numero * 2         # entrega el resultado a quien llamó

doble = duplicar(8)           # doble vale 16
print(doble + 1)              # 17: se puede seguir operando
print(duplicar(duplicar(3)))  # 12: se puede usar dentro de otra llamada
```

| `print(...)` dentro de la función | `return ...` |
|---|---|
| **muestra** el valor en pantalla | **entrega** el valor a quien llamó |
| el resultado **no** se puede guardar ni operar | el resultado se guarda en una variable o se usa en una expresión |
| la función devuelve `None` | la función devuelve ese valor |

> - `return` **termina** la función: lo que esté después **no se ejecuta**.
> - Una función sin `return` devuelve `None`.
