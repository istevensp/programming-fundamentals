## 1. Leer funciones (a mano, sin ejecutar)

Una función es una "máquina": recibe datos por sus **parámetros**, hace su trabajo con **sus propias variables** y **devuelve** un resultado con `return`. Para entender una función hay que poder seguirla **a mano**.

| Concepto | Qué significa |
|---|---|
| **parámetro** | nombre que aparece en el `def` (`def doble(n)`: `n`) |
| **argumento** | valor que se pasa al llamarla (`doble(5)`: `5`) |
| `return` | **devuelve** un valor a quien llamó; la función termina ahí |
| sin `return` | la función devuelve `None` |
| `print` | **muestra** en pantalla, pero no devuelve nada |
| variable local | existe solo dentro de la función; cambiarla no afecta a las de afuera |

En esta sección **no ejecutes** las funciones: el código está solo en el texto. Usa los valores de **tus datos** (los que mostró la celda 🔧) y escribe el **resultado final** (por ejemplo `17`, no `CANTIDAD * 2 + 7`).
